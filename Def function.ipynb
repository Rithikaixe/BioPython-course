```python
def find_snp(a,b):
    for i in range(len(a)):
        if a[i] != b[i]:
            print ("SNP at position", i + 1, ":", a[i], "to", b[i])
                   
```

```python
a = "ATGC"
b = "ATAC"

find_snp(a,b)
```

```python
a = "ATTGATAGCGTAGTAGCT"
b = "ATTAATAGCATAGTAGCT"

find_snp(a,b)
```
